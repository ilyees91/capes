Le tableau initial vient de la photo. Le code suit maintenant la précision du professeur : le tri modifie la liste sans return tab. Il corrige aussi la borne de recherche du minimum : le préfixe déjà placé doit être exclu.

```python
def indice_minimum(tab, i):
    # Précondition : 0 <= i < len(tab).
    # Renvoie un indice du minimum parmi les cases i à len(tab)-1.
    imin = i
    for k in range(i + 1, len(tab)):
        if tab[k] < tab[imin]:
            imin = k
    return imin


def tri_selection(tab):
    for i in range(len(tab) - 1):
        imin = indice_minimum(tab, i)
        tab[i], tab[imin] = tab[imin], tab[i]
```

## Le premier piège : chercher au bon endroit

Après le premier échange, la valeur 2 est déjà placée. Pour remplir la case 1, le minimum recherché est 3, d'indice 5. Une recherche sur tout le tableau retrouverait 2 et casserait le préfixe déjà trié.

L'auxiliaire renvoie un **indice**, pour pouvoir échanger les cases. Le tri modifie la liste reçue sur place.

```python
apres_premier_tour = [2, 8, 7, 9, 5, 3, 4]
imin = indice_minimum(apres_premier_tour, 1)
assert imin == 5
print("Indice :", imin, "; valeur :", apres_premier_tour[imin])

exemple = [8, 2, 7, 9, 5, 3, 4]
resultat = tri_selection(exemple)
assert resultat is None
print("Tableau trié :", exemple)
```

## Reproduire les étapes du schéma

La cellule suivante reprend la boucle du tri et affiche son état après chaque échange. Les copies servent uniquement à conserver l'historique pour l'étude ; elles ne font pas partie du tri en espace constant.

```python
tab = [8, 2, 7, 9, 5, 3, 4]
etapes = [tab.copy()]
for i in range(len(tab) - 1):
    imin = indice_minimum(tab, i)
    print(f"i={i} : minimum {tab[imin]}, indice {imin}")
    tab[i], tab[imin] = tab[imin], tab[i]
    etapes.append(tab.copy())
    print("  Après échange :", tab)

assert etapes == [
    [8, 2, 7, 9, 5, 3, 4],
    [2, 8, 7, 9, 5, 3, 4],
    [2, 3, 7, 9, 5, 8, 4],
    [2, 3, 4, 9, 5, 8, 7],
    [2, 3, 4, 5, 9, 8, 7],
    [2, 3, 4, 5, 7, 8, 9],
    [2, 3, 4, 5, 7, 8, 9],
]
```

## Résumer la preuve et le coût

Avant le tour `i`, le préfixe est trié, ses valeurs sont inférieures ou égales à celles de la zone restante, et le tableau conserve tous les éléments initiaux. Placer un minimum de la zone restante prolonge ce préfixe d'une case. Les boucles sont finies.

Les recherches effectuent `(n - 1) + (n - 2) + ... + 1 = n(n - 1)/2` comparaisons : **21 pour 7 éléments**, et Θ(n²) en général, même sur une liste déjà triée. L'algorithme de tri utilise O(1) espace supplémentaire ; le stockage des étapes ci-dessus est seulement un outil de visualisation.

## Vérifications

On teste les listes vides, les singletons, les répétitions, les valeurs négatives et différents ordres. Le résultat doit contenir toutes les valeurs initiales, dans l'ordre croissant.

```python
from itertools import product

nombre = 0
for taille in range(6):
    for valeurs in product((-1, 0, 1), repeat=taille):
        tab = list(valeurs)
        resultat = tri_selection(tab)
        assert resultat is None
        assert tab == sorted(valeurs)
        nombre += 1

assert indice_minimum([2, 8, 7, 9, 5, 3, 4], 1) == 5
assert indice_minimum([3, 1, 1], 1) == 1
assert indice_minimum([7], 0) == 0
assert sum(range(7)) == 21
print(nombre, "tris vérifiés ; recherche du minimum vérifiée sur les cas limites.")
```
