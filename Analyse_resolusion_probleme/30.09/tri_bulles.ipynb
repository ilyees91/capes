![Premier passage du tri à bulles](tri_bulles.svg)

La photo montre les deux boucles et la comparaison. La partie droite de l'échange, coupée sur l'image, est complétée par l'échange simultané standard. La fonction modifie la liste et renvoie implicitement `None`. Les traces et tests ci-dessous sont des compléments.

```python
def tri_a_bulles(valeurs):
    n = len(valeurs)
    for i in range(n):
        for j in range(0, n - i - 1):
            if valeurs[j] > valeurs[j + 1]:
                valeurs[j], valeurs[j + 1] = valeurs[j + 1], valeurs[j]
```

## Utiliser le tri sur place

On appelle la fonction, puis on lit la liste. Le résultat renvoyé n'est pas une nouvelle liste triée.

```python
exemple = [8, 3, 5, 2, 7, 4]
retour = tri_a_bulles(exemple)
assert retour is None
assert exemple == [2, 3, 4, 5, 7, 8]
print("Liste triée :", exemple)
print("Retour :", retour)
```

## Suivre les passages

À chaque passage, un maximum de la zone active rejoint le suffixe déjà placé. Le premier passage correspond au schéma. Le dernier tour de `range(n)` ne fait aucune comparaison.

Les compteurs et les copies servent à observer le programme ; ils ne font pas partie du code minimal du professeur.

```python
valeurs = [8, 3, 5, 2, 7, 4]
n = len(valeurs)
comparaisons = 0
echanges = 0
premier_passage = [valeurs.copy()]
apres_passages = []
for i in range(n):
    for j in range(0, n - i - 1):
        comparaisons += 1
        if valeurs[j] > valeurs[j + 1]:
            valeurs[j], valeurs[j + 1] = valeurs[j + 1], valeurs[j]
            echanges += 1
        if i == 0:
            premier_passage.append(valeurs.copy())
    apres_passages.append(valeurs.copy())
    print(f"Après le passage {i} : {valeurs}")

assert premier_passage == [
    [8, 3, 5, 2, 7, 4],
    [3, 8, 5, 2, 7, 4],
    [3, 5, 8, 2, 7, 4],
    [3, 5, 2, 8, 7, 4],
    [3, 5, 2, 7, 8, 4],
    [3, 5, 2, 7, 4, 8],
]
assert apres_passages == [
    [3, 5, 2, 7, 4, 8],
    [3, 2, 5, 4, 7, 8],
    [2, 3, 4, 5, 7, 8],
    [2, 3, 4, 5, 7, 8],
    [2, 3, 4, 5, 7, 8],
    [2, 3, 4, 5, 7, 8],
]
assert comparaisons == 15 and echanges == 9
print(comparaisons, "comparaisons et", echanges, "échanges.")
```

## Preuve et complexité

Avant le tour intérieur `j`, la case `j` contient un maximum du préfixe allant de 0 à `j`. Après la comparaison des deux voisins, ce maximum est en `j + 1`. Il atteint ainsi la fin de la zone active.

Avant le passage `i`, les `i` dernières valeurs sont triées et supérieures ou égales à celles de la zone restante. Le nouveau maximum rejoint ce suffixe à gauche. Les échanges conservent les éléments. Les deux `range` sont finis : l'algorithme termine et trie correctement.

**Ce code sans arrêt anticipé effectue toujours `n(n - 1)/2` comparaisons**, soit un temps Θ(n²), même si l'entrée est triée. L'espace auxiliaire est O(1). Avec `>`, les éléments de même clé conservent leur ordre : le tri est stable.

Un arrêt après un passage sans échange donnerait un meilleur cas Θ(n), mais cette optimisation n'est pas dans le code photographié.

```python
from itertools import product

nombre = 0
for taille in range(6):
    for valeurs_initiales in product((-1, 0, 1), repeat=taille):
        tab = list(valeurs_initiales)
        assert tri_a_bulles(tab) is None
        assert tab == sorted(valeurs_initiales)
        nombre += 1

class Element:
    def __init__(self, cle, nom):
        self.cle = cle
        self.nom = nom

    def __gt__(self, autre):
        return self.cle > autre.cle

objets = [Element(2, "A"), Element(1, "B"), Element(2, "C"), Element(1, "D")]
tri_a_bulles(objets)
assert [x.nom for x in objets] == ["B", "D", "A", "C"]
print(nombre, "tris vérifiés ; retour None et stabilité vérifiés.")
```
