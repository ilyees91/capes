# 1. VARIANTS — Prouver la terminaison

Pour chaque fonction : préciser les préconditions, choisir un variant, montrer qu'il diminue strictement et qu'il reste minoré tant que l'exécution continue.

Pour un variant entier, une borne inférieure et une décroissance **stricte** sont nécessaires : `n - s` peut être négatif après le dernier tour, à condition d'être minoré à chaque test où la boucle continue. Dans une présentation exigeant un naturel à tous les tests, on peut adapter le variant ; un exemple apparaît avec la racine carrée.

Les sept fonctions du TD sont définies ici. Exécuter leurs cellules permet ensuite de lancer les exemples et tests de la deuxième partie.

## 1.1 `facto(n)`

````python
def facto(n):
    f = 1
    while n > 0:
        f = f * n
        n = n - 1
    return f
````

**Précondition :** `n` est naturel. Notons `N` sa valeur initiale.

**Au tableau :** variant `n`, strictement décroissant puisque `n` devient `n - 1`, et positif sous la condition du `while`.

**Rédaction complète :** initialement, `n = N >= 0`. Si le test `n > 0` est vrai, le nouvel entier `n - 1` reste naturel et est strictement inférieur à `n`. Il ne peut donc pas y avoir une infinité de tours. Pour `N = 0`, aucun tour n'est exécuté et le résultat est `1`, conformément à `0! = 1`.

## 1.2 `programmeMystere1(tab)`

````python
def programmeMystere1(tab):
    tab2 = [0] * len(tab)
    j = 0
    k = len(tab2) - 1
    for i in range(len(tab)):
        if tab[i] >= 0:
            tab2[j] = tab[i]
            j += 1
        else:
            tab2[k] = tab[i]
            k -= 1
    return tab2
````

**Précondition :** une liste de nombres comparables à zéro.

**Variant :** avec `N = len(tab)`, le nombre d'éléments restant à traiter vaut `N - i` au début du tour d'indice `i`. Il diminue de 1. La boucle `for` parcourt un intervalle fini.

`i` augmente : ce n'est donc pas lui le variant décroissant. Pour une liste vide, aucun tour n'est exécuté.

La règle sur `for` concerne ici un **itérateur fini** et un corps qui termine. Elle ne veut pas dire que toute boucle `for` Python termine : on pourrait fournir un itérateur infini ou placer une boucle infinie dans son corps.

## 1.3 `racineApprochee(n)`

````python
def racineApprochee(n):
    c = 0
    s = 0
    while s <= n:
        s = s + 2 * c + 1
        c = c + 1
    return c - 1
````

**Précondition :** `n` est un entier naturel.

**Variant :** `n - s` est entier positif ou nul tant que `s <= n`. Dans le code corrigé par le professeur, `s` augmente de `2*c + 1 >= 1` (avec l'ancien `c`), donc le variant décroît strictement. Il peut devenir négatif au dernier tour, lorsque la condition devient fausse : la boucle termine.

Pour un variant naturel y compris à la sortie, choisir `max(0, n - s + 1)`, strictement décroissant à chaque tour exécuté.

La correction du PDF consiste à calculer `s` avant d'incrémenter `c`, puis à renvoyer `c - 1`. La preuve avec l'invariant `s = c²` est en partie 2.3.

## 1.4 `programmeMystere2(tab)`

````python
def programmeMystere2(tab):
    i = 0
    j = len(tab) - 1
    while i < j:
        if tab[i] < tab[j]:
            i = i + 1
        else:
            j = j - 1
    return tab[i]
````

**Précondition :** `tab` est non vide et contient des valeurs comparables, par exemple des entiers.

**Variant :** `j - i`. Il est positif pendant la boucle et diminue de 1 à chaque tour, quel que soit le branchement.

À chaque tour, soit `i` augmente de 1, soit `j` diminue de 1. Ils ne changent pas tous les deux au même tour.

## 1.5 `division_euclidienne(a, b)`

````python
def division_euclidienne(a, b):
    q = 0
    r = a
    while r >= b:
        r = r - b
        q = q + 1
    return (q, r)
````

**Préconditions :** `a` naturel et `b` entier strictement positif.

**Variant :** `r`. Il reste naturel, car on soustrait `b` seulement si `r >= b`. Il diminue strictement puisque `b > 0` : `r' = r - b < r`. Une suite de naturels strictement décroissante est finie, donc la boucle termine.

**Pourquoi imposer `b > 0` ?** Avec `b = 0`, le reste ne changerait jamais et le test resterait vrai pour `a >= 0`. Si `a < b`, aucun tour n'est nécessaire : quotient `0`, reste `a`.

## 1.6 `pgcd(a, b)`

````python
def pgcd(a, b):
    if b == 0:
        return a
    else:
        return pgcd(b, a % b)
````

Le symbole de reste du PDF est retranscrit par l'opérateur Python `%`.

**Préconditions :** `a` et `b` naturels ; on les suppose non tous deux nuls pour le PGCD usuel.

**Variant :** le second argument `b`. Pour `b > 0`, on a `0 <= a % b < b`. Le variant reste naturel et diminue strictement. Quand il atteint zéro, la fonction renvoie immédiatement `a`.

Exemple : les seconds arguments de `pgcd(30, 18)` sont `18`, `12`, `6`, puis `0`. On ne choisit pas forcément le premier argument : dans `pgcd(3, 8)`, il passe d'abord de `3` à `8`, donc il augmente.

## 1.7 `ackermann(m, n)`

````python
def ackermann(m, n):
    if m == 0:
        return n + 1
    elif n == 0:
        return ackermann(m - 1, 1)
    else:
        return ackermann(m - 1, ackermann(m, n - 1))
````

**Préconditions :** `m` et `n` sont des entiers naturels. Les deux paramètres ne diminuent pas nécessairement ensemble.

### Ce que le professeur a écrit

**Au tableau :** le couple **`(m, n)` est un variant**. Les comparaisons indiquées sont :

$$
(m-1,1) < (m,n),
$$

$$
(m,n-1) < (m,n),
$$

$$
\bigl(m-1,\operatorname{ackermann}(m,n-1)\bigr) < (m,n).
$$

Le tableau indique également le caractère **bien fondé** de l'ordre sur les couples et une démonstration par récurrence bien fondée.

### Complément — quel ordre utilise-t-on ?

Il s'agit de l'**ordre lexicographique sur les couples de naturels** : on compare d'abord la première coordonnée, puis la seconde en cas d'égalité.

$$
(a,b)<_{\mathrm{lex}}(m,n)
\iff a<m\ \text{ou}\ (a=m\ \text{et}\ b<n).
$$

Ainsi, `(1, 1000) < (2, 3)` : la première coordonnée est prioritaire. La seconde peut augmenter si la première diminue.

Cet ordre est bien fondé : dans une chaîne décroissante, la première coordonnée ne peut diminuer qu'un nombre fini de fois. Si la chaîne était infinie, cette coordonnée finirait par être constante, et la seconde devrait alors diminuer indéfiniment dans les naturels, ce qui est impossible.

### Preuve de terminaison, étape par étape

On prouve par récurrence bien fondée que **chaque appel termine et renvoie un entier naturel**. Supposons cette propriété établie pour tous les couples strictement inférieurs à `(m, n)`.

| Cas | Appel à justifier | Pourquoi le couple diminue |
| --- | --- | --- |
| `m == 0` | Aucun appel récursif | Retour immédiat de `n + 1` |
| `m > 0`, `n == 0` | `ackermann(m - 1, 1)` | La première coordonnée diminue |
| `m > 0`, `n > 0` — appel intérieur | `ackermann(m, n - 1)` | Première coordonnée identique, seconde plus petite |
| Même cas — appel extérieur | `ackermann(m - 1, résultat)` | La première coordonnée diminue, quel que soit le résultat naturel |

Dans le dernier cas, Python doit d'abord calculer l'appel intérieur. L'hypothèse de récurrence garantit qu'il termine et produit un naturel `r`. On peut alors appliquer cette même hypothèse au couple `(m - 1, r)`.

> **À retenir :** il ne suffit pas de regarder le `m - 1` de l'appel extérieur ; il faut aussi justifier que l'appel intérieur termine.

### Un petit exemple

```text
ackermann(1, 1)
= ackermann(0, ackermann(1, 0))
= ackermann(0, ackermann(0, 1))
= ackermann(0, 2)
= 3
```

La terminaison ne signifie pas que le calcul est rapide. Cette fonction croît très vite ; pour les essais Python, rester sur de petites entrées. La somme `m + n` n'est pas le variant retenu, car la seconde coordonnée peut fortement augmenter.

### Complément — Pourquoi Ackermann est-elle un exemple marquant ?

Cette fonction illustre la différence entre **terminer en théorie** et **pouvoir calculer un résultat en pratique**. La preuve par ordre lexicographique garantit la terminaison pour tous les couples de naturels, mais les résultats et le nombre d'appels deviennent très grands très vite. En Python, la profondeur de récursion peut être dépassée avant d'obtenir le résultat.

Par exemple, pour la définition du TD, `ackermann(3, n) = 2**(n + 3) - 3`. Dès que le premier argument augmente, la croissance devient encore beaucoup plus rapide. On se limite donc à de petits appels pour les essais.

Ce complément explique le phénomène mathématique ; les détails historiques de l'anecdote du professeur ne figurent pas dans les photos fournies.

## 2.1 `facto(n)`

**Objectif :** renvoyer la factorielle de la valeur initiale `N`.

**Invariant au test de boucle :**

$$f \times n! = N! \qquad\text{et}\qquad 0 \le n \le N.$$

- **Initialisation :** `f = 1` et `n = N`.
- **Conservation :** pour `n > 0`, les nouvelles valeurs sont `f' = f × n` et `n' = n - 1`. Ainsi `f' × n'! = f × n! = N!`.
- **Sortie :** `n = 0`, donc `f = N!` puisque `0! = 1`.

| Étape pour `N = 4` | `n` | `f` |
| --- | --- | --- |
| Départ | 4 | 1 |
| 1 | 3 | 4 |
| 2 | 2 | 12 |
| 3 | 1 | 24 |
| 4 — sortie | 0 | 24 |

## 2.2 `programmeMystere1(tab)`

### L'invariant du professeur — reformulé avec des bornes explicites

La photo décrit les deux zones remplies de `tab2` : les éléments positifs ou nuls dans leur ordre d'apparition, puis les éléments strictement négatifs dans l'ordre inverse. Les bornes manuscrites sont peu lisibles ; on les explicite ci-dessous en notation Python, cohérente avec le code.

**Au début de l'itération d'indice `i`, avant de traiter `tab[i]` :**

> `tab2[:j]` contient exactement les éléments positifs ou nuls de `tab[:i]`, dans leur ordre d'apparition ; `tab2[k+1:]` contient exactement les éléments strictement négatifs de `tab[:i]`, dans l'ordre inverse de leur apparition.

Les répétitions sont conservées. On note `N = len(tab)` et on complète la propriété par :

```text
0 <= j <= k + 1 <= N
j + (N - 1 - k) = i
```

Autrement dit, le nombre de cases remplies à gauche plus le nombre de cases remplies à droite est égal au nombre d'éléments déjà examinés.

### Lire les indices sans se tromper

| Zone | Indices inclus | Contenu |
| --- | --- | --- |
| `tab[:i]` | 0 à `i - 1` | Éléments déjà examinés |
| `tab2[:j]` | 0 à `j - 1` | Positifs ou nuls, dans l'ordre d'apparition |
| `tab2[j:k+1]` | `j` à `k` | Cases restant à remplir |
| `tab2[k+1:]` | `k + 1` à `N - 1` | Négatifs, dans l'ordre inverse |

`j` désigne la prochaine case à remplir à gauche ; `k`, la prochaine à remplir à droite. **La zone négative commence donc à `k + 1`, pas à `k`.**

Si l'on formule la propriété **à la fin du tour d'indice `i`**, l'élément `tab[i]` est déjà traité : il faut remplacer `tab[:i]` par `tab[:i+1]` et le nombre d'éléments traités par `i + 1`.

### Exemple du schéma

Pour `tab = [-3, 8, -2, 7, 9]`, après avoir traité `-3` et `8`, on est au début du tour `i = 2` :

```text
tab  = [-3, 8 | -2, 7, 9]
         lus | à examiner

tab2 = [8 | 0, 0, 0 | -3]
       gauche  libre   droite

i = 2, j = 1, k = 3
```

Les zéros du milieu sont les valeurs d'initialisation de cases encore libres ; ils ne représentent pas des éléments déjà copiés.

| Élément traité | `tab2` après le tour | `j` | `k` |
| --- | --- | --- | --- |
| `-3` | `[0, 0, 0, 0, -3]` | 0 | 3 |
| `8` | `[8, 0, 0, 0, -3]` | 1 | 3 |
| `-2` | `[8, 0, 0, -2, -3]` | 1 | 2 |
| `7` | `[8, 7, 0, -2, -3]` | 2 | 2 |
| `9` | `[8, 7, 9, -2, -3]` | 3 | 2 |

Les négatifs étaient `-3`, puis `-2` dans l'entrée ; ils sont `-2`, puis `-3` dans la sortie parce qu'on remplit la partie droite de droite à gauche.

### Preuve de l'invariant

**1. Initialisation.** Avant tout traitement, le préfixe lu est vide. `j = 0` et `k = N - 1` : les deux zones remplies sont vides, et `j + (N - 1 - k) = 0`.

**2. Conservation.** Supposons la propriété vraie avant le tour `i`. Comme `i < N`, l'égalité sur le nombre de cases remplies donne `k - j + 1 = N - i > 0`. Il reste donc au moins une case libre et les indices utilisés sont valides.

- Si `tab[i] >= 0`, on l'écrit en case `j`, puis on augmente `j`. Il est ajouté à la fin de la zone positive ou nulle : l'ordre d'apparition est conservé.
- Sinon, on l'écrit en case `k`, puis on diminue `k`. Il est ajouté devant les négatifs précédents : l'ordre inverse est conservé.

Dans les deux cas, une seule case libre est remplie et un élément supplémentaire est traité. La propriété est donc vraie pour le préfixe `tab[:i+1]`.

**3. Conclusion.** Après les `N` tours, tous les éléments ont été examinés. L'égalité devient `j + (N - 1 - k) = N`, donc `j = k + 1` : il ne reste aucune case libre. La sortie contient tous les éléments positifs ou nuls, puis tous les négatifs, avec les ordres annoncés.

Pour la preuve, on raisonne sur le nombre d'éléments traités : la variable Python `i` ne prend pas la valeur `N` à la sortie du `for` (et n'est pas définie par cette boucle si elle est vide).

**Résultat :** une séparation selon le signe, pas un tri numérique. La liste vide est également correctement traitée.

````python
print(programmeMystere1([-2, 4, -1, 0, 3]))
# [4, 0, 3, -1, -2]
````

## 2.3 `racineApprochee(n)` — Correction du professeur

**Précondition :** `n` est un entier naturel. Le code du [TD avec la racine corrigée](TD2_erreur_corrige.pdf) et la cellule de définition en partie 1 ont été mis à jour. [PDF original conservé](TD2_original.pdf).

### Les deux corrections du tableau

1. Dans la boucle, calculer **`s = s + 2*c + 1` avant `c = c + 1`**.
2. À la sortie, renvoyer **`c - 1`**, et non `c`.

### Invariant écrit par le professeur

$$s = c^2.$$

On observe cette propriété au test de boucle, avant chaque tour et à la sortie.

**Initialisation.** `s = 0` et `c = 0`, donc `s = c²` est vrai.

**Conservation.** Supposons `s = c²` avant un tour. En notant `s'` et `c'` les nouvelles valeurs :

$$s' = s + 2c + 1 = c^2 + 2c + 1 = (c+1)^2,$$

puis `c' = c + 1`. On obtient donc `s' = (c')²` : la propriété est conservée.

**L'ordre des instructions est essentiel :** le `c` utilisé dans `2*c + 1` est l'ancien compteur.

### À la sortie — raisonnement du professeur

Notons $S_c$ la valeur de `s` lorsque le compteur vaut `c`. Comme `n >= 0`, on effectue au moins un tour. Le dernier test vrai et le premier test faux donnent :

$$S_{c-1} \le n < S_c.$$

Avec l'invariant :

$$(c-1)^2 \le n < c^2.$$

Puisque `c - 1 >= 0`, on peut prendre les racines :

$$c-1 \le \sqrt{n} < c.$$

Ainsi :

$$\boxed{c-1 = \left\lfloor\sqrt{n}\right\rfloor}.$$

La fonction renvoie donc la **partie entière de la racine carrée de `n`**. C'est pourquoi il faut renvoyer `c - 1`.

### Exemple : `n = 8`

| Moment du test | `c` | `s = c²` | `s <= 8` |
| --- | --- | --- | --- |
| Départ | 0 | 0 | Vrai |
| Après un tour | 1 | 1 | Vrai |
| Après deux tours | 2 | 4 | Vrai |
| Après trois tours | 3 | 9 | Faux : sortie |

Le résultat est `3 - 1 = 2`. Pour `n = 9`, on effectue un tour supplémentaire : `c = 4`, `s = 16`, et on renvoie `3`.

**Complément :** la preuve de terminaison reste fondée sur `n - s` sous la condition de boucle. Avec le code corrigé, `s` augmente de `2*c + 1 >= 1`, où `c` est sa valeur avant le tour. Si l'on souhaite un variant naturel même à la sortie, on peut utiliser `max(0, n - s + 1)`.

**Comprendre l'erreur de l'original.** Si l'on augmente d'abord `c`, alors on ajoute successivement `3`, `5`, `7`, etc. : l'invariant devient `s = c(c + 2)`, et non `s = c²`. Par exemple, pour `n = 8`, le code original renvoie `3` alors que la partie entière de la racine vaut `2`. Cet ancien invariant décrit correctement le code fautif, mais ne prouve pas le résultat attendu.

**Cas limite `n = 0` :** un tour fait passer `(c, s)` de `(0, 0)` à `(1, 1)` ; la fonction renvoie `0`.

## 2.4 `programmeMystere2(tab)` — Maximum d'un tableau

**Précondition :** le tableau est non vide et ses éléments sont comparables, par exemple des entiers.

### Invariant du professeur

> Le maximum du tableau se situe entre les indices `i` et `j`, compris.

Plus précisément, **au moins une occurrence du maximum** reste dans cet intervalle. Il peut y avoir plusieurs occurrences du maximum dans le tableau.

En notation mathématique, au test de boucle :

$$0 \le i \le j < N \quad\text{et}\quad \max(tab[i:j+1]) = \max(tab),$$

avec `N = len(tab)`. La tranche Python `tab[i:j+1]` inclut bien la case `j`.

### 1. Initialisation — au tableau

Au départ, `i = 0` et `j = len(tab) - 1`. L'intervalle est donc le tableau entier : il contient nécessairement son maximum. L'invariant est vrai avant le premier tour.

### 2. Hérédité — au tableau, avec les deux cas explicités

Supposons l'invariant vrai au début d'un tour. Le professeur explique que l'on élimine le plus petit des deux éléments aux extrémités, `tab[i]` et `tab[j]`.

- **Si `tab[i] < tab[j]`**, on augmente `i` de 1. La valeur écartée est strictement inférieure à une valeur que l'on conserve : elle ne peut pas être le maximum de l'intervalle. Le maximum reste dans le nouvel intervalle.
- **Sinon, `tab[i] >= tab[j]`**, on diminue `j` de 1. La valeur conservée en `i` est au moins aussi grande que celle écartée en `j`. Il reste donc une occurrence du maximum dans le nouvel intervalle.

**Cas d'égalité :** si les deux extrémités valent le maximum, on peut en écarter une puisque l'autre reste. C'est pourquoi on parle d'« au moins une occurrence ».

La propriété est conservée à chaque tour. On ne supprime pas physiquement les éléments : on réduit seulement l'intervalle examiné ; `tab` reste inchangé.

### 3. Conclusion à la sortie — complément

La terminaison a été prouvée en partie 1 grâce au variant `j - i`.

À la sortie, `i < j` est faux, donc `i >= j`. Or l'invariant donne `i <= j` : on a donc **`i = j`**.

L'intervalle contient une seule case. Puisqu'il contient encore un maximum du tableau, **`tab[i]` est le maximum recherché**. C'est la valeur renvoyée par la fonction.

### Exemple du professeur : `[4, 3, 7, 9, 8, 5, 6]`

Les indices ci-dessous sont ceux du début de chaque tour. Les valeurs à l'extérieur de l'intervalle ne sont plus examinées.

| Tour | `i` | `j` | Comparaison | Mise à jour | Intervalle après le tour |
| --- | --- | --- | --- | --- | --- |
| 1 | 0 | 6 | `4 < 6` : vrai | `i = 1` | `[3, 7, 9, 8, 5, 6]` |
| 2 | 1 | 6 | `3 < 6` : vrai | `i = 2` | `[7, 9, 8, 5, 6]` |
| 3 | 2 | 6 | `7 < 6` : faux | `j = 5` | `[7, 9, 8, 5]` |
| 4 | 2 | 5 | `7 < 5` : faux | `j = 4` | `[7, 9, 8]` |
| 5 | 2 | 4 | `7 < 8` : vrai | `i = 3` | `[9, 8]` |
| 6 | 3 | 4 | `9 < 8` : faux | `j = 3` | `[9]` |

À chaque ligne, le maximum `9` reste dans l'intervalle. À la sortie, `i = j = 3`, et la fonction renvoie `tab[3]`, donc **9**.

**À retenir :** on n'a pas besoin de savoir où se trouve le maximum pour éliminer une extrémité dont on sait qu'elle n'est pas indispensable.


````python
print(programmeMystere2([3, 9, 2, 9, 4]))  # 9
````

Sur une liste vide, la boucle ne s'exécute pas, mais `tab[i]` provoque une `IndexError`. Terminer la boucle ne suffit donc pas à garantir un résultat valide pour toutes les entrées.

## 2.5 `division_euclidienne(a, b)`

**Objectif :** pour `a >= 0` et `b > 0`, renvoyer des naturels `(q, r)` tels que `a = b*q + r` et `0 <= r < b`.

**Invariant au test de boucle :** `a = b*q + r`, avec `q >= 0` et `r >= 0`.

1. **Initialisation :** `q = 0` et `r = a >= 0`, donc `a = b*0 + a`.
2. **Conservation :** un tour est exécuté seulement si `r >= b`. Les nouvelles valeurs `q' = q+1` et `r' = r-b` restent naturelles et vérifient `b*q' + r' = b*(q+1) + r-b = b*q+r = a`.
3. **Sortie :** le test est faux, donc `r < b`. Avec l'invariant, cela donne exactement la division euclidienne de `a` par `b`.

| Étape pour `(a, b) = (17, 5)` | `q` | `r` | Égalité conservée |
| --- | --- | --- | --- |
| Départ | 0 | 17 | `17 = 5*0 + 17` |
| Après un tour | 1 | 12 | `17 = 5*1 + 12` |
| Après deux tours | 2 | 7 | `17 = 5*2 + 7` |
| Sortie | 3 | 2 | `17 = 5*3 + 2`, avec `0 <= 2 < 5` |

La boucle fait exactement `a // b` soustractions. La preuve du variant et cette preuve de l'invariant établissent ensemble la correction totale.

````python
print(division_euclidienne(17, 5))  # (3, 2)
````

## 2.6 `pgcd(a, b)` — Correction par récurrence

**Propriété :** pour des naturels `a` et `b` non tous deux nuls, la fonction renvoie leur PGCD.

Il n'y a pas de boucle : on raisonne par récurrence forte sur le second argument `b`.

- **Cas de base :** pour `b = 0`, la fonction renvoie `a`, qui est le PGCD de `a` et `0`.
- **Hérédité :** pour `b > 0`, le reste `r = a % b` vérifie `0 <= r < b`. L'hypothèse de récurrence assure que `pgcd(b, r)` renvoie le PGCD de `b` et `r`.
- **Propriété conservée entre les appels :** `a` et `b` ont les mêmes diviseurs communs que `b` et `r`, puisque `a = q*b + r`. En effet, si un entier divise `a` et `b`, il divise `r = a-q*b` ; réciproquement, s'il divise `b` et `r`, il divise `a = q*b+r`. Donc leur PGCD est identique.

On peut parler d'une propriété conservée le long des appels, mais pas d'un invariant de boucle.

```text
pgcd(30, 18) → pgcd(18, 12) → pgcd(12, 6) → pgcd(6, 0) → 6
```

## 2.7 `ackermann(m, n)` — Correction par récurrence bien fondée

**Objectif :** calculer la fonction mathématique `A` définie par les trois équations :

$$A(0,n)=n+1,$$

$$A(m,0)=A(m-1,1)\quad\text{pour }m>0,$$

$$A(m,n)=A(m-1,A(m,n-1))\quad\text{pour }m>0,\ n>0.$$

Il n'y a pas d'invariant de boucle à chercher. On prouve que le programme renvoie `A(m, n)` par récurrence bien fondée sur `(m, n)`, avec l'ordre lexicographique défini en première partie.

- **Si `m = 0` :** le programme renvoie `n + 1`, conformément à la première équation.
- **Si `m > 0` et `n = 0` :** le couple `(m - 1, 1)` est plus petit. L'hypothèse de récurrence garantit le résultat `A(m - 1, 1)`, donc `A(m, 0)`.
- **Si `m > 0` et `n > 0` :** l'appel intérieur porte sur `(m, n - 1)`, plus petit ; il renvoie donc `r = A(m, n - 1)`, un naturel. L'appel extérieur porte alors sur `(m - 1, r)`, lui aussi plus petit, et renvoie `A(m - 1, r) = A(m, n)`.

Chaque branche respecte la définition mathématique. Avec la terminaison établie en première partie, on obtient la correction totale dans le modèle mathématique d'exécution.

# 3. Exercice 2 — Concevoir les algorithmes et prouver leur résultat

Les programmes ci-dessous sont des **propositions de correction complémentaires** fondées sur les cinq questions du PDF. On fixe leur code avant d'écrire les invariants : changer le sens d'un parcours peut changer la propriété à démontrer.

Pour les sommes, on travaille ici sur des entiers, afin que les égalités soient exactes. Pour les minimums et les tris, on suppose des valeurs totalement ordonnées, par exemple des entiers. Les trois tris modifient la liste reçue et la renvoient ; ils acceptent aussi la liste vide.

Une preuve de tri doit montrer deux choses : **les valeurs sont ordonnées** et **les éléments de l'entrée sont conservés, avec leurs répétitions**. Dans les preuves, `A` désigne le contenu initial de la liste, sans imposer une copie dans le programme.

## 3.1 Somme d'un tableau

**Idée :** additionner les valeurs une par une. La somme d'une liste vide vaut `0`.

````python
def somme(tab):
    s = 0
    for i in range(len(tab)):
        s += tab[i]
    return s
````

**Invariant :** avant le tour d'indice `i`, `s = sum(tab[:i])`.

- **Initialisation :** pour `i = 0`, le préfixe est vide et `s = 0`.
- **Conservation :** ajouter `tab[i]` transforme la somme des `i` premières valeurs en celle des `i+1` premières valeurs.
- **Sortie :** après `N = len(tab)` tours, `s = sum(tab)`.

**Terminaison :** le nombre de valeurs restant à traiter, `N - i`, diminue de 1 entre deux tours ; le parcours est fini. Pour `[4, -2, 7]`, les sommes successives sont `0`, `4`, `2`, puis `9`.

## 3.2 Minimum d'un tableau

**Précondition mathématique :** le tableau est non vide. On initialise le minimum avec la première valeur, car choisir arbitrairement `0` donnerait une mauvaise réponse sur `[5, 8]`. Le code signale explicitement une liste vide.

````python
def minimum(tab):
    if not tab:
        raise ValueError("Le minimum exige une liste non vide.")
    m = tab[0]
    for i in range(1, len(tab)):
        if tab[i] < m:
            m = tab[i]
    return m
````

**Invariant :** avant le tour d'indice `i`, `m = min(tab[:i])`, avec `1 <= i <= N`.

- **Initialisation :** `m = tab[0]` est le minimum du préfixe de longueur 1.
- **Conservation :** après la comparaison, `m` est le plus petit entre l'ancien minimum et `tab[i]`, donc le minimum du préfixe de longueur `i+1`.
- **Sortie :** tous les éléments ont été examinés, donc `m = min(tab)`.

**Terminaison :** le parcours des `N-1` valeurs restantes est fini. Pour une liste d'un seul élément, aucun tour n'est nécessaire.

## 3.3 Tri par sélection

**Idée :** à la position `i`, chercher un minimum dans la zone restante `tab[i:]`, puis le placer en `i` par un échange.

````python
def tri_selection(tab):
    N = len(tab)
    for i in range(N):
        indice_min = i
        for j in range(i + 1, N):
            if tab[j] < tab[indice_min]:
                indice_min = j
        tab[i], tab[indice_min] = tab[indice_min], tab[i]
    return tab
````

**Invariant de la boucle extérieure :** avant le tour `i`, `tab[:i]` est trié, chaque élément de ce préfixe est inférieur ou égal à chaque élément de `tab[i:]`, et `tab` contient les mêmes éléments que `A`.

**Boucle intérieure :** avant le tour `j`, `indice_min` est l'indice d'un minimum de `tab[i:j]`. Au départ, ce segment contient seulement `tab[i]`. La comparaison avec `tab[j]` étend la recherche d'une case. Après le parcours, `indice_min` désigne donc un minimum de `tab[i:]`.

**Preuve extérieure :**

1. Au départ, le préfixe vide est trié et rien n'a changé dans le tableau.
2. L'échange place en `i` un minimum de la zone restante. Il est au moins aussi grand que les éléments du préfixe déjà placé et au plus aussi grand que les valeurs encore à droite : le préfixe agrandi vérifie la même propriété. Un échange conserve tous les éléments.
3. Après `N` tours, le préfixe est le tableau entier : il est trié et conserve l'entrée.

**Terminaison :** les deux parcours sont finis ; pour des indices de tours successifs, les quantités `N-i` et `N-j` diminuent. Chaque parcours intérieur doit terminer avant de pouvoir passer au tour extérieur suivant.

Exemple : `[3, 1, 2] → [1, 3, 2] → [1, 2, 3]`.

## 3.4 Tri par insertion

**Idée :** le préfixe est déjà trié ; on insère la valeur suivante à sa place en la faisant reculer par échanges avec son voisin gauche. Cette variante par échanges facilite la preuve de conservation des éléments.

````python
def tri_insertion(tab):
    for i in range(1, len(tab)):
        j = i
        while j > 0 and tab[j] < tab[j - 1]:
            tab[j - 1], tab[j] = tab[j], tab[j - 1]
            j -= 1
    return tab
````

**Invariant extérieur :** avant le tour `i`, `tab[:i]` est trié et contient les mêmes éléments que `A[:i]` ; la partie à partir de `i` n'a pas encore été modifiée.

Au départ, un préfixe de longueur 1 est trié. Pour insérer la nouvelle valeur, notons `x` la valeur située en `i` au début de ce tour. Dans la boucle intérieure, `x` occupe toujours la case `j`.

**Invariant intérieur :** `0 <= j <= i`, `tab[j] = x` ; le préfixe `tab[:i+1]` privé de cette case est trié, et tous les éléments de `tab[j+1:i+1]` sont strictement supérieurs à `x`. Les échanges conservent les éléments.

- Initialement, retirer `x` laisse le préfixe déjà trié ; la zone située après `x`, jusqu'à `i`, est vide.
- Si le voisin gauche est strictement supérieur à `x`, l'échange déplace `x` d'une case vers la gauche. Les autres valeurs restent dans le même ordre et la valeur dépassée rejoint la zone supérieure à `x`.
- À la sortie, `j = 0` ou le voisin gauche est inférieur ou égal à `x`. Les valeurs à gauche sont donc toutes inférieures ou égales à `x`, celles à droite lui sont supérieures : le préfixe de longueur `i+1` est trié.

L'invariant extérieur est conservé ; une fois tous les éléments insérés, la liste entière est triée et contient les mêmes éléments que l'entrée.

**Terminaison :** le parcours extérieur est fini. Dans chaque boucle intérieure, `j` est naturel et diminue de 1 ; c'est un variant. Le test `j > 0` est évalué avant `tab[j-1]`, ce qui évite de comparer avec la dernière case lorsque `j = 0`.

Exemple : après avoir trié le préfixe `[2, 4]`, l'insertion de `3` transforme `[2, 4, 3]` en `[2, 3, 4]`.

## 3.5 Tri à bulles

**Idée :** comparer des voisins de gauche à droite et échanger s'ils sont dans le mauvais ordre. Après un passage jusqu'à `fin`, un maximum de la zone active est placé en `fin`.

````python
def tri_bulles(tab):
    for fin in range(len(tab) - 1, 0, -1):
        for j in range(fin):
            if tab[j] > tab[j + 1]:
                tab[j], tab[j + 1] = tab[j + 1], tab[j]
    return tab
````

**Invariant extérieur :** au début d'un passage, `tab[fin+1:]` est trié et tous ses éléments sont supérieurs ou égaux à ceux de la zone active `tab[:fin+1]`. Le tableau reste une permutation de l'entrée.

**Invariant intérieur :** avant la comparaison d'indice `j`, `tab[j]` est un maximum de `tab[:j+1]`. Après la comparaison et l'éventuel échange, un maximum de `tab[:j+2]` se trouve en `j+1`. Au dernier échange éventuel, un maximum de la zone active se trouve donc en `fin`.

**Initialisation :** le suffixe déjà placé est vide. **Conservation :** le maximum nouvellement placé est inférieur ou égal aux valeurs du suffixe précédent ; il l'agrandit d'une case tout en le laissant trié. Les échanges conservent les éléments. **Sortie :** pour une liste de longueur au moins 2, il ne reste qu'une case active, nécessairement inférieure ou égale à toutes les suivantes : la liste entière est triée. Les listes de longueur 0 ou 1 le sont déjà.

**Terminaison :** les deux `range` sont finis. Le nombre de passages restant à réaliser est `fin`, qui diminue de 1 ; pendant un passage, `fin-j` diminue de 1.

Exemple de premier passage : `[3, 1, 2] → [1, 3, 2] → [1, 2, 3]`. Le dernier `3` est alors définitivement placé.

### Comparer les trois tris proposés

On note `N` la longueur et on compte les comparaisons/échanges comme des opérations élémentaires.

| Algorithme | Temps dans le pire cas | Meilleur cas de ce code | Espace supplémentaire |
| --- | --- | --- | --- |
| Sélection | O(N²) | O(N²) | O(1) |
| Insertion | O(N²) | O(N) si la liste est déjà triée | O(1) |
| Bulles | O(N²) | O(N²), sans arrêt anticipé | O(1) |

Le tri à bulles pourrait s'arrêter après un passage sans échange ; cette optimisation n'est pas utilisée ici. Les invariants ci-dessus décrivent exactement les programmes proposés.

# 4. Exercice 3 — Classer les élèves à partir de l'invariant

Le PDF travaille sur des **objets élèves**, munis d'une méthode `niveau()`. Les valeurs autorisées sont **2 = seconde**, **1 = première**, **0 = terminale**. On veut cet ordre de groupes : `2`, puis `1`, puis `0`.

La proposition ci-dessous suit l'invariant donné dans l'énoncé. Pour le visualiser, on représente un élève par son niveau ; le programme échange bien les objets complets.

```text
[ secondes : 2 ][ premières : 1 ][ non traités : ? ][ terminales : 0 ]
0               i                j                 k+1              N
```

| Zone | Propriété |
| --- | --- |
| `tab[:i]` | Élèves de seconde |
| `tab[i:j]` | Élèves de première |
| `tab[j:k+1]` | Élèves restant à examiner |
| `tab[k+1:]` | Élèves de terminale |

**Invariant complet :** les quatre zones ont ces propriétés, `0 <= i <= j <= k+1 <= N`, et le tableau contient exactement les mêmes élèves que l'entrée. La propriété est observée au test de boucle, donc aussi après chaque tour.

**Préconditions :** chaque élément possède une méthode `niveau()` qui termine et renvoie toujours, pour cet élève, une valeur parmi `0`, `1`, `2`.

````python
def classer_eleves(tab):
    i = 0
    j = 0
    k = len(tab) - 1
    while j <= k:
        niveau = tab[j].niveau()
        if niveau == 2:
            tab[i], tab[j] = tab[j], tab[i]
            i += 1
            j += 1
        elif niveau == 1:
            j += 1
        elif niveau == 0:
            tab[j], tab[k] = tab[k], tab[j]
            k -= 1
        else:
            raise ValueError("Un niveau doit valoir 0, 1 ou 2.")
    return tab
````

## 4.1 Initialisation et conservation

Initialement, `i = j = 0` et `k = N-1`. Les trois zones classées sont vides, la zone inconnue est le tableau entier, et les bornes sont correctes, même si `N = 0`.

Supposons l'invariant vrai avant un tour. Puisque `j <= k`, la zone inconnue n'est pas vide et la case `j` est valide.

- **Niveau 2 :** on échange la case `j` avec la case `i`. Si `i < j`, l'élève qui était en `i` est une première et se retrouve en `j` ; avancer les deux indices agrandit la zone des secondes et conserve celle des premières. Si `i = j`, l'échange se fait avec la même case et ne pose aucun problème.
- **Niveau 1 :** avancer seulement `j` agrandit d'une case la zone des premières.
- **Niveau 0 :** l'échange avec `k` place une terminale à droite ; reculer `k` l'intègre à la zone classée. On garde `j` au même endroit, car l'élève arrivé depuis la case `k` était encore inconnu. Si `j = k`, cet échange avec lui-même vide la zone inconnue.

Dans chaque cas, les bornes restent ordonnées, les zones ont les propriétés attendues et les échanges préservent tous les élèves. Le classement ne garantit pas de conserver l'ordre relatif des élèves d'un même niveau : ce n'est pas un tri stable.

## 4.2 Terminaison et correction à la sortie

**Variant :** `V = k-j+1`, le nombre d'élèves non traités. Il est naturel au test, strictement positif si le tour est exécuté, et diminue exactement de 1 dans chaque branche : `j` augmente ou `k` diminue.

La boucle termine donc après exactement `N` tours sous les préconditions. À la sortie, `j > k` ; l'invariant donne `j <= k+1`, donc `j = k+1`. La zone inconnue est vide, et les trois zones restantes donnent le classement demandé. Le temps est O(N) si `niveau()` prend un temps constant, et l'espace supplémentaire est O(1).

## 4.3 Exemple commenté

On affiche seulement les niveaux pour suivre les échanges.

| Départ | `[1, 0, 2, 1, 0, 2]` | 0 | 0 | 5 | 6 |
| 1 | `[1, 0, 2, 1, 0, 2]` | 0 | 1 | 5 | 5 |

La petite classe suivante sert uniquement à fabriquer des élèves pour l'exemple ; l'algorithme accepte toute classe conforme à la méthode demandée par l'énoncé.

````python
class Eleve:
    def __init__(self, nom, niveau):
        self.nom = nom
        self._niveau = niveau

    def niveau(self):
        return self._niveau


eleves = [Eleve(f"Élève {i}", niveau)
          for i, niveau in enumerate([1, 0, 2, 1, 0, 2])]
classer_eleves(eleves)
print([eleve.niveau() for eleve in eleves])  # [2, 2, 1, 1, 0, 0]
````
